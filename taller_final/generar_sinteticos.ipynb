```python
import pandas as pd
import numpy as np
import random

# Configurar el generador de números aleatorios
np.random.seed(42)
random.seed(42)

# Definir el número de registros
num_records = 10000

# Generar datos sintéticos ajustados del primer ejemplo
data = {
    'ProductoID': np.random.randint(1, 101, num_records),
    'Categoria': np.random.choice(['Camisetas', 'Pantalonetas', 'Busos'], num_records),
    'Precio': np.round(np.random.uniform(25000, 85000, num_records)),
    'Estacion': np.random.choice(['Primavera', 'Verano', 'Otoño', 'Invierno'], num_records),
    'Promocion': np.random.choice([0, 1], num_records),  # 0: No promoción, 1: Promoción
    'VentasHistoricas': np.random.randint(150, 900, num_records),
    'Demanda': np.random.randint(30, 600, num_records)
}

# Crear un DataFrame
df_primero = pd.DataFrame(data)

# Generar datos adicionales para el segundo ejemplo
# Lógica ajustada:
# - VentasDurantePromocion aumentadas si hay promoción
# - VentasPostPromocion menores a VentasDurantePromocion
# - InventarioInicial basado en la Demanda proyectada y el Inventario Inicial ajustado

df_primero['VentasDurantePromocion'] = np.where(df_primero['Promocion'] == 1,
                                                df_primero['VentasHistoricas'] * np.random.uniform(1.2, 1.5, len(df_primero)),
                                                df_primero['VentasHistoricas'] * np.random.uniform(0.8, 1.1, len(df_primero)))
                                                
df_primero['VentasPostPromocion'] = df_primero['VentasDurantePromocion'] * np.random.uniform(0.5, 0.8, len(df_primero))

df_primero['InventarioInicial'] = df_primero['Demanda'] * np.random.uniform(1.1, 1.3, len(df_primero))

# Calcular el Inventario Final
df_primero['InventarioFinal'] = df_primero['InventarioInicial'] - (df_primero['VentasDurantePromocion'] + df_primero['VentasPostPromocion'])

# Asegurar que el inventario final no sea negativo
df_primero['InventarioFinal'] = df_primero['InventarioFinal'].apply(lambda x: max(x, 0))

# Guardar el DataFrame actualizado en un archivo CSV
df_primero.to_csv('promociones_productos.csv', index=False, encoding='utf-8-sig')

print("Base de datos sintética generada y guardada como 'promociones_productos.csv'")

```

```python
import pandas as pd
import numpy as np
import random

# Configurar el generador de números aleatorios
np.random.seed(42)
random.seed(42)

# Cargar los datos del segundo ejemplo
df_primero = pd.read_csv('promociones_productos.csv')

# Generar datos adicionales para el tercer ejemplo
# - InventarioRecepcion: Cantidad de productos recibidos inicialmente
# - InventarioAlmacenamiento: Cantidad después de la recepción y antes de las ventas
# - InventarioVentas: Cantidad de productos restantes después de las ventas

df_primero['InventarioRecepcion'] = df_primero['InventarioInicial'] * np.random.uniform(0.95, 1.05, len(df_primero))
df_primero['InventarioAlmacenamiento'] = df_primero['InventarioRecepcion'] - df_primero['VentasDurantePromocion'] * np.random.uniform(0.95, 1.05, len(df_primero))
df_primero['InventarioVentas'] = df_primero['InventarioAlmacenamiento'] - df_primero['VentasPostPromocion'] * np.random.uniform(0.95, 1.05, len(df_primero))

# Asegurar que ningún valor de inventario sea negativo
df_primero['InventarioAlmacenamiento'] = df_primero['InventarioAlmacenamiento'].apply(lambda x: max(x, 0))
df_primero['InventarioVentas'] = df_primero['InventarioVentas'].apply(lambda x: max(x, 0))

# Guardar el DataFrame actualizado en un archivo CSV
df_primero.to_csv('datos_rfid_productos.csv', index=False, encoding='utf-8-sig')

print("Base de datos sintética generada y guardada como 'datos_rfid_productos.csv'")

```
