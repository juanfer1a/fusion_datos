```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
import numpy as np

# Cargar los resultados del Nivel 2
resultados_nivel2 = pd.read_csv('resultados_dempster_shafer.csv')

# Definir las características (X) y la variable objetivo (y) para el Nivel 3
# Supongamos que la variable objetivo es una valoración del riesgo o impacto, generamos datos sintéticos para esta variable
resultados_nivel2['Riesgo'] = np.random.randint(1, 11, len(resultados_nivel2))

# Convertir las características a un DataFrame
X = resultados_nivel2[['Creencia', 'Plausibilidad', 'Probabilidad_Pignistica']]
y = resultados_nivel2['Riesgo']

# Normalizar las características
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Dividir los datos en conjuntos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
# Definir el modelo de bosque aleatorio
forest = RandomForestRegressor(random_state=42)

# Ajustar el modelo
forest.fit(X_train, y_train)

# Hacer predicciones sobre el conjunto de prueba
y_pred = forest.predict(X_test)

# Evaluar el modelo
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print(f"Error cuadrático medio (MSE): {mse}")
print(f"Raíz del error cuadrático medio (RMSE): {rmse}")

# Guardar las predicciones en un archivo CSV
predicciones_riesgo = pd.DataFrame({'Riesgo_Predicho': y_pred})
predicciones_riesgo.to_csv('predicciones_riesgo.csv', index=False)

print("Predicciones de riesgo guardadas en 'predicciones_riesgo.csv'")
```
